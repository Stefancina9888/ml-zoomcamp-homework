Q5. Max fuel efficiency

```python
print("Max fuel efficiency (cars from asia):")
asian_cars = df[df["origin"] == "Asia"]
print(asian_cars["fuel_efficiency_mpg"].max())

# Maximum fuel efficiency for cars from Asia is 41.2
```

Q6. Median value of horsepower

```python
print("Median value of horsepower:")
horsepower_median = df["horsepower"].median()
print(horsepower_median)

print("Most frequent horsepower:")
most_frequent_horsepower = df["horsepower"].value_counts().idxmax()
print(most_frequent_horsepower)

print("Remaining NaN horsepower values")
df = df.fillna({"horsepower": most_frequent_horsepower})
print(df.horsepower.isnull().sum())

print("New median value of horsepower:")
new_horsepower_median = df["horsepower"].median()
print(new_horsepower_median)

# Median value of horsepower has changed and decreased.
```

Q7. Sum of weights

```python
cars_from_asia = df[df["origin"] == "Asia"]
asia_model_weight = cars_from_asia[["model_year","vehicle_weight"]]
asia_model_weight = asia_model_weight.head(7)
print(asia_model_weight)

X = asia_model_weight.values
XTX = X.T.dot(X)

XTX_INV = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_INV.dot(X.T).dot(y)

print("Sum of all elements in w:")
print(w.sum())

# The sum of all elements in w is 0.369
```
